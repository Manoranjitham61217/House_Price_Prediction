```python
hp = hp.fillna(hp.median(numeric_only=True))
```

```python
hp = hp[hp['price'] > 0]
```

```python
median_price = hp['price'].median()
hp['price_class'] = (hp['price'] > median_price).astype(int)
```

```python
hp.head()
```

```python
x=hp.drop(['price','price_class'],axis=1)
y = (hp['price_class'])
```

```python
x = x.select_dtypes(include=['int64','float64'])
```

```python
print(x.shape)
print(y.shape)
```

```python
from sklearn.feature_selection import SelectKBest,f_regression
selector=SelectKBest(f_regression,k=10)
x_selected=selector.fit_transform(x,y)
selected_features=x.columns[selector.get_support()]
```

```python
selected_features
```

```python
x.corr()
```

```python
top_10=x.corr().head(11)[1:].index[:10]
corr_matrix = hp[list(top_10) + ['price']].corr()
```

```python
plt.Figure(figsize=(10,6))
sns.heatmap(corr_matrix,annot=True,cmap='cividis',fmt='.2f')
plt.title("Correlation of each Features")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x_selected)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_scaled,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from xgboost import XGBClassifier
import time
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,roc_curve, auc
start=time.time()
xgb_model = XGBClassifier(
    n_estimators=600,
    learning_rate=0.05,
    max_depth=8,
    subsample=0.8,
    colsample_bytree=0.8,
    gamma=1.0,
    reg_lambda=1.0,
    random_state=42,
    objective="reg:squarederror",
)

xgb_model.fit(x_train, y_train)
end=time.time()
y_pred=xgb_model.predict(x_test)
accuracy=accuracy_score(y_test,y_pred)
report=classification_report(y_test,y_pred)
print("Accuracy:  ",accuracy)
print("Classification Report:\n ",report)
print("TIME: ",end-start)
```

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("Confusion Matrix - XGBoost Price Classification")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
